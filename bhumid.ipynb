#polynomial operations  


```python
import numpy as np
x=np.poly1d([5,4,7 ])
y=np.poly1d([7,8,5 ])
z=np.polyadd(x,y)#addition
print(z)
z= np.polymul(x,y)#multiplication
print(z)
```
