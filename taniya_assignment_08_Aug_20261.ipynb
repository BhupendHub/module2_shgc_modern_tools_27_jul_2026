```python
A=np.array([ [1,2,], [3,4] ])
B=np.array([ [5,6], [7,8] ])
C=A=B
print("Matrix Addition:\n",C)
D=A@B
D=np.dot(A,B)
print("Matrix Multiplication:\n",D)
E=np.linalg.inv(A)
print("Matrix Inverse:\n",E)
```

```python
p1=np.poly1d([2,4,6])
p2=np.poly1d([1,3,5])
p3=p1+p2
p3=np.polyadd(p1,p2)
print("Polynomial Addition:\n",p3)
p4=p1*p2
p3=np.polymul(p1,p2)
print("Polymial Multiplication:\n",p4)
```

## **Working with Matplotlib - drawing bar chart for playerVs Runs**

```python
import matplotlib.pyplot as plt
names=["kirti","nisha","priya"]
runs=[90,80,50]
plt.bar(names,runs,color='r')
plt.title("player vs Runs")
plt.xlabel("player")
plt.grid(False)
plt.ylabel("Runs")
plt.show()
```
